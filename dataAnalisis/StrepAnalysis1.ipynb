# Growth rate

```python
#import experimental data
data1 = np.genfromtxt('coel.csv', delimiter=',', skip_header=0) #0:Minutes, 1:Length, 2:Foci\
data1[:,0] = data1[:,0] * (1/60) #convert minutes to hours

data2 = np.genfromtxt('vene.csv', delimiter=',', skip_header=0) #0:Minutes, 1:Length, 2:Foci
data2[:,0] = data2[:,0] * (1/60) #convert minutes to hours



initial=0#int(4*(60/1)*(60/1)*(1/5))
initialExp = 30 #25

# Fit a line to the log of the average data
slope1, intercept1 = np.polyfit(data1[initialExp:, 0], np.log(data1[initialExp:, 1]), 1) #time(hours) vs log(length)
slope2, intercept2 = np.polyfit(data2[initialExp:, 0], np.log(data2[initialExp:, 1]), 1) #time(hours)  vs log(length)

x = np.linspace(5, 10, 10000)
y = np.e ** (slope1*x + intercept1)

x2 = np.linspace(5, 10, 10000)
y2 = np.e ** (slope2*x2 + intercept2)

plt.figure( figsize=(10, 8))

plt.plot(data1[0:, 0], data1[0:, 1], linestyle=':', label='Experiment 1', color='red', linewidth=2)
plt.plot(data2[0:, 0], data2[0:, 1], linestyle=':', label='Experiment 2', color='blue', linewidth=2)




# Initialize lists to store the data
time_data = []
log_data = []
slopes = []
intercepts = []
# Loop through all the files

initial=int(3*(60/1)*(60/1)*(1/5))

for i in range(0, 5):
    # Generate the file name
    file_name = f'Simu({i}).csv'
    
    # Load the data
    data = np.genfromtxt(file_name, delimiter=',', skip_header=0)
    
    # Transform the data
    data[:, 0] = data[:, 0] * (1/60) * (1/60)  # Convert seconds to hours
   

    # Fit a line to the data
    slope, intercept = np.polyfit(data[initial:, 0], np.log(data[initial:, 1]), 1)
    
    # Store the transformed data
    time_data.append(data[0:, 0])
    log_data.append(data[0:, 1])
    slopes.append(slope)
    intercepts.append(intercept)



# Calculate the average slope and intercept
average_slope = np.mean(slopes)
average_intercept = np.mean(intercepts)

# Plot the average line
t = np.linspace(3, 8, 10000)  # 10 hours
u = np.e ** (average_slope*t + average_intercept)   

# Plot number of branches over time

# Plot each dataset
for i in range(0, 5):
    plt.plot(time_data[i]+2, log_data[i], linestyle=':', color='gray', linewidth=1)

plt.plot(time_data[0]+2, log_data[0], linestyle=':', color='black', linewidth=0.7, label='Individual simulations')
plt.plot(x, y, color='red', linewidth=1.5, label=r'Experimental growth rate 1 = {:.2f} $h^{{-1}}$'.format(slope1) )
plt.plot(x2, y2,   color='blue', linewidth=1.5, label=r'Experimental growth rate 2 = {:.2f} $h^{{-1}}$'.format(slope2) )


plt.plot(t+2, u, color='black', linewidth=1.5, label=r'Simulated growth rate = {:.2f} $h^{{-1}}$'.format(average_slope))


plt.tick_params(axis='both', which='major', labelsize=12)
plt.legend(fontsize=12)
plt.xlabel('Time (hours)', fontsize=14)
plt.ylabel('Length [μm]', fontsize=14)
plt.yscale('log')
plt.xlim(5, 10)



```

```python
import numpy as np
import matplotlib.pyplot as plt

# Initialize lists to store the data
time_data = []
log_data = []
slopes = []
intercepts = []

initial=int(4*(60/1)*(60/1)*(1/5))

# Loop through all the files
for i in range(10, 133):
    # Generate the file name
    file_name = f'Simulation({i}).csv'
    
    # Load the data
    data = np.genfromtxt(file_name, delimiter=',', skip_header=0)
    
    # Transform the data
    data[:, 0] = data[:, 0] * (1/60) * (1/60)  # Convert seconds to hours
    data[:, 2] = np.log(data[:, 2])

    # Fit a line to the data
    slope, intercept = np.polyfit(data[initial:, 0], data[initial:, 2], 1)
    
    # Store the transformed data
    time_data.append(data[:, 0])
    log_data.append(data[:, 2])
    slopes.append(slope)
    intercepts.append(intercept)


t= np.linspace(4, 10, 10000) # 10 hours

average_slope = np.mean(slopes)
average_intercept = np.mean(intercepts)

# Plot number of branches over time
plt.figure(figsize=(10, 8))

# Plot each dataset
for i in range(10, 133):
    plt.scatter(time_data[i-10], log_data[i-10], s=0.5)

plt.plot(t, average_slope*t + average_intercept, color='black', linewidth=1, label='Average slope = {:.4f}'.format(average_slope))

plt.xlabel('Time (hours)')
plt.ylabel(' ln(Number of branches)')
plt.legend()
# Add legend and show plot
plt.show()


```
