```python
import numpy as np 
import matplotlib.pyplot as plt
# Import dataset Stack_spots.csv
data1 = np.genfromtxt('Stack_spots.csv', delimiter=',', skip_header=0, dtype=str)

col_names = data1[0]
#errase first 4 rows

data1 = data1[4:]
unique_IDs1 = np.unique(data1[:,2])

```

```python
# Let's get the points for the first ID

distances1 = []

for i in range(len(unique_IDs1)):
    ID1 = data1[data1[:,2] == unique_IDs1[i]]
    x = ID1[:,4].astype(float)
    y = ID1[:,5].astype(float)
    t = ID1[:,8].astype(float)

    #sort the points by time
    sort_index = np.argsort(t)
    x = x[sort_index]
    y = -y[sort_index]
    t = t[sort_index]

    #calculate the distances between points
    for j in range(len(x)-1):
        distances1.append(np.sqrt((x[j+1]-x[j])**2 + (y[j+1]-y[j])**2))
    

    plt.plot(x,y)

plt.show()

#plot histogram of distances
plt.hist(distances1, bins=100)
plt.show()

    
```

### Exponential? No!


```python
import numpy as np

# Set the scale parameter (lambda) for the exponential distribution
lambda_ = 1.999

# Sample 1000 elements from the exponential distribution
samples = np.random.exponential(1/lambda_, 10000)

# Display the sampled elements
plt.hist(samples, bins=100, density=True)

lambda_mleTest = 1/np.mean(distancesTotal)  
x = np.linspace(0, np.max(samples), 1000)
pdfTest = lambda_mleTest * np.exp(-lambda_mleTest * x)
plt.plot(x, pdfTest, label='MLE')
plt.show()
```

```python
from scipy.stats import expon, kstest
```

```python
# Fit an Exponential distribution
lambda_mle = 1 / np.mean(distancesTotal)  # MLE for exponential
x = np.linspace(0, max(distancesTotal), 100)
pdf = lambda_mle * np.exp(-lambda_mle * x)

# Plot the PDF over the histogram
plt.plot(x, pdf, 'r-', label=f'Exponential fit ($\lambda$={lambda_mle:.2f})')

# compare the fit to the data by plotting the data and the fit together
plt.hist(distancesTotal, bins=100, density=True, label='Data')
plt.legend()

plt.show()


# Perfom a test to check if the data is exponentially distributed
test=kstest(distancesTotal, expon.cdf, args=(0, 1/lambda_mle))
print(test)

if test[1] > 0.05:
    print('The data is exponentially distributed')
    print('The p-value is:', test[1])
else:
    print('The data is not exponentially distributed')
    print('The p-value is:', test[1])

```

### Weibull distribution? No!


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import weibull_min


# 2. Plot the histogram of the data
plt.hist(distancesTotal, bins=100, density=True, alpha=0.6, color='b', label='Histogram of samples')

# 3. Fit the Weibull distribution to the data
fitted_shape, loc, fitted_scale = weibull_min.fit(distancesTotal)

# 4. Plot the fitted Weibull PDF
x = np.linspace(0, max(distancesTotal), 1000)
pdf_fitted = weibull_min.pdf(x, fitted_shape, loc, fitted_scale)
plt.plot(x, pdf_fitted, 'r-', label=f'Fitted Weibull PDF (k={fitted_shape:.2f}, λ={fitted_scale:.2f})')

# 5. Add labels and legend
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Weibull Distribution: Histogram and Fitted PDF')
plt.legend()

# 6. Show the plot
plt.show()


from scipy.stats import weibull_min, probplot

#generate QQ plot
fig, ax = plt.subplots()
probplot(distancesTotal, dist=weibull_min, sparams=(fitted_shape, loc, fitted_scale), plot=ax)
plt.show()

from scipy.stats import kstest

# Fit the Weibull distribution to the data
fitted_shape, loc, fitted_scale = weibull_min.fit(distancesTotal)

# Perform the K-S test
D, p_value = kstest(distancesTotal, 'weibull_min', args=(fitted_shape, loc, fitted_scale))

print(f"K-S test statistic: {D}")
print(f"P-value: {p_value}")

if p_value > 0.05:
    print("Fail to reject the null hypothesis: The data may come from a Weibull distribution.")
else:
    print("Reject the null hypothesis: The data does not come from a Weibull distribution.")
```

## Gamma? No

```python
#check if there are values in distancesTotal that are negative
print(np.min(distancesTotal))

#check if there are values in distancesTotal that are not finite
print(np.sum(np.isnan(distancesTotal)))

#add a small positive value to all distances
epsilon = 1e-15
distancesTotal = distancesTotal + epsilon
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma

# Fit the Gamma distribution to your data
shape, loc, scale = gamma.fit(distancesTotal, floc=0)  # Fix location to 0 for simplicity

# Plot the histogram and Gamma PDF
plt.hist(distancesTotal, bins=100, density=True, alpha=0.6, color='b', label='Histogram of samples')
x = np.linspace(0, max(distancesTotal), 1000)
pdf_fitted = gamma.pdf(x, shape, loc, scale)
plt.plot(x, pdf_fitted, 'r-', label=f'Gamma PDF (k={shape:.2f}, θ={scale:.2f})')
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Gamma Distribution: Histogram and Fitted PDF')
plt.legend()
plt.show()

# Perform a K-S test
from scipy.stats import kstest
ks_stat, p_value = kstest(distancesTotal, 'gamma', args=(shape, loc, scale))
print(f"K-S Statistic: {ks_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value > 0.05:
    print("Fail to reject the null hypothesis: The data may come from a Gamma distribution.")
else:
    print("Reject the null hypothesis: The data does not come from a Gamma distribution.")


```

Lognormal? yes

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm, kstest

# Fit the Lognormal distribution to your data
shape, loc, scale = lognorm.fit(distancesTotal)

# Plot the histogram and Lognormal PDF
plt.hist(distancesTotal, bins=200, density=True, alpha=0.6, color='b', label='Histogram of foci displacements at 10-minute intervals')
x = np.linspace(0, max(distancesTotal), 1000)
pdf_fitted = lognorm.pdf(x, shape, loc, scale)
plt.plot(x, pdf_fitted, 'r-', label=f'Lognormal PDF (σ={shape:.2f})')

# Perform a K-S test
ks_stat, p_value = kstest(distancesTotal, 'lognorm', args=(shape, loc, scale))
print(f"K-S Statistic: {ks_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Determine the test result
if p_value > 0.05:
    test_result = "K-S test fails to reject the null hypothesis. \nThe data may come from a Lognormal distribution."
else:
    test_result = "Reject the null hypothesis: The data does not come from a Lognormal distribution."

# Add labels, title, legend, and test result text
plt.xlabel('Foci displacement from previous frame (µm)')
plt.ylabel('Density')
plt.title('Lognormal Distribution: Histogram and Fitted PDF')
plt.legend()

# Add the test result text to the plot
plt.text(0.1, 0.95, test_result, transform=plt.gca().transAxes, fontsize=10, verticalalignment='top')

# Show the plot
plt.show()

print(np.mean(distancesTotal))
```

```python
#read data from simulation
SimData = np.genfromtxt('Traces(2).csv', delimiter=',', skip_header=0, dtype=str)

#filter all the tracks for the first id
uniqueIDsSim = np.unique(SimData[:,0])

distancesSim = []

for i in range(len(uniqueIDsSim)):
    ID = SimData[SimData[:,0] == uniqueIDsSim[i]]
    x = ID[:,2].astype(float)*0.1
    y = ID[:,3].astype(float)*0.1
    t = ID[:,1].astype(float)

    #sort the points by time
    sort_index = np.argsort(t)
    x = x[sort_index]
    y = -y[sort_index]
    t = t[sort_index]

    #calculate the distances between points
    for j in range(len(x)-1):
        distancesSim.append(np.sqrt((x[j+1]-x[j])**2 + (y[j+1]-y[j])**2))

plt.hist(distancesSim, bins=100, density=True)

#Fit the Lognormal distribution to the data
shape, loc, scale = lognorm.fit(distancesSim)

# Perform a K-S test to see if it is lognormal
ks_stat, p_value = kstest(distancesSim, 'lognorm', args=(shape, loc, scale))
print(f"K-S Statistic: {ks_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value > 0.05:
    test_result = "K-S test fails to reject the null hypothesis. \nThe data may come from a Lognormal distribution."
else:
    test_result = "Reject the null hypothesis: The data does not come from a Lognormal distribution."
print(test_result)
# fit a lognormal distribution to the data and plot the pdf on top of the histogram



# Plot the  Lognormal PDF

x = np.linspace(0, max(distancesSim), 1000)
pdf_fitted = lognorm.pdf(x, shape, loc, scale)

plt.plot(x, pdf_fitted, 'r-', label=f'Lognormal PDF (σ={shape:.2f})')

print(np.mean(distancesSim))

plt.show()
```
