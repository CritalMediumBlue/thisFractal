```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Function to load simulation data and convert time to hours
def load_simulation_data(folder, sim_indices):
 
    simulations = []
    for idx in sim_indices:
        data = np.genfromtxt(f'./inputdata/{folder}/Simulation({idx}).csv', 
                             delimiter=',', skip_header=0)
        data[:, 0] = data[:, 0] * (1/60) * (1/60)  # Convert seconds to hours
        simulations.append(data)
    return simulations


# Load simulation data from normal conditions
sim_indices1 = np.arange(0, 13)  # Simulations 0 to 13
normal_sims = load_simulation_data('normalConditions', sim_indices1)
initialTime=2 #hours
initial=int(initialTime*(60/1)*(60/1)*(1/5)) 
slopes_length_normal = []
slopes_branch_normal = []
intercepts_length_normal = []
intercepts_branch_normal = []

for sim in normal_sims:
     # Fit a line to the length
    slopeLength, interceptLength = np.polyfit(sim[initial:, 0], np.log(sim[initial:, 1]), 1)
    # Fit a line to the branches
    slopeBranch, interceptBranch = np.polyfit(sim[initial:, 0], np.log(sim[initial:, 2]), 1)
    slopes_length_normal.append(slopeLength)
    slopes_branch_normal.append(slopeBranch)
    intercepts_length_normal.append(interceptLength)
    intercepts_branch_normal.append(interceptBranch)

Mean_slope_length_normal = np.mean(slopes_length_normal)
sd_slope_length_normal = np.std(slopes_length_normal)

Mean_slope_branch_normal = np.mean(slopes_branch_normal)
sd_slope_branch_normal = np.std(slopes_branch_normal)

Mean_intercept_length_normal = np.mean(intercepts_length_normal)
sd_intercept_length_normal = np.std(intercepts_length_normal)

Mean_intercept_branch_normal = np.mean(intercepts_branch_normal)
sd_intercept_branch_normal = np.std(intercepts_branch_normal)




# Load simulation data from no binding conditions
sim_indices2 = np.arange(0, 16)  # Simulations 0 to 16
no_binding_sims = load_simulation_data('noBinding', sim_indices2)
slopes_length_no_binding = []
slopes_branch_no_binding = []
intercepts_length_no_binding = []
intercepts_branch_no_binding = []
for sim in no_binding_sims:
     # Fit a line to the length
    slopeLength, interceptLength = np.polyfit(sim[initial:, 0], np.log(sim[initial:, 1]), 1)
    # Fit a line to the branches
    slopeBranch, interceptBranch = np.polyfit(sim[initial:, 0], np.log(sim[initial:, 2]), 1)
    slopes_length_no_binding.append(slopeLength)
    slopes_branch_no_binding.append(slopeBranch)
    intercepts_length_no_binding.append(interceptLength)
    intercepts_branch_no_binding.append(interceptBranch)

Mean_slope_length_no_binding = np.mean(slopes_length_no_binding)
sd_slope_length_no_binding = np.std(slopes_length_no_binding)

Mean_slope_branch_no_binding = np.mean(slopes_branch_no_binding)
sd_slope_branch_no_binding = np.std(slopes_branch_no_binding)

Mean_intercept_length_no_binding = np.mean(intercepts_length_no_binding)
sd_intercept_length_no_binding = np.std(intercepts_length_no_binding)

Mean_intercept_branch_no_binding = np.mean(intercepts_branch_no_binding)
sd_intercept_branch_no_binding = np.std(intercepts_branch_no_binding)


# Load simulation data from non-local conditions
sim_indices3 = np.arange(0, 24)  # Simulations 0 to 24
non_local = load_simulation_data('nonLocal', sim_indices3)

slopes_length_non_local = []
slopes_branch_non_local = []
intercepts_length_non_local = []
intercepts_branch_non_local = []
for sim in non_local:
     # Fit a line to the length
    slopeLength, interceptLength = np.polyfit(sim[initial:, 0], np.log(sim[initial:, 1]), 1)
    # Fit a line to the branches
    slopeBranch, interceptBranch = np.polyfit(sim[initial:, 0], np.log(sim[initial:, 2]), 1)
    slopes_length_non_local.append(slopeLength)
    slopes_branch_non_local.append(slopeBranch)
    intercepts_length_non_local.append(interceptLength)
    intercepts_branch_non_local.append(interceptBranch)

Mean_slope_length_non_local = np.mean(slopes_length_non_local)
sd_slope_length_non_local = np.std(slopes_length_non_local)

Mean_slope_branch_non_local = np.mean(slopes_branch_non_local)
sd_slope_branch_non_local = np.std(slopes_branch_non_local)

Mean_intercept_length_non_local = np.mean(intercepts_length_non_local)
sd_intercept_length_non_local = np.std(intercepts_length_non_local)

Mean_intercept_branch_non_local = np.mean(intercepts_branch_non_local)
sd_intercept_branch_non_local = np.std(intercepts_branch_non_local)


# Plot Length vs Time ######### Linear scale #########
color_normal = 'blue'
color_no_binding = 'red'
color_non_local = 'green'


plt.figure(figsize=(10,6))

# Plot normal conditions
for i, sim in enumerate(normal_sims):
    label = 'Normal Conditions' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_normal, linestyle=':', linewidth=0.5)

# Plot no binding conditions
for i, sim in enumerate(no_binding_sims):
    label = 'No TIPOC Binding' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_no_binding, linestyle=':', linewidth=0.5)

# Plot non-local conditions
for i, sim in enumerate(non_local):
    label = 'No Local production' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_non_local, linestyle=':', linewidth=0.5)


plt.xlabel('Time (hours)')
plt.ylabel('Length (µm)')
plt.title('Total Hyphae Length Over Time')
plt.grid()
plt.legend()
plt.savefig('./outputdata/streptomyces_growth.png', dpi=300)
plt.show()
plt.clf()




# Plot Length vs Time ######### logarithmic scale #########
color_normal = 'blue'
color_no_binding = 'red'
color_non_local = 'green'


plt.figure(figsize=(10,6))

# Plot normal conditions
for i, sim in enumerate(normal_sims):
    label = 'Normal Conditions' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_normal, linestyle=':', linewidth=0.5)

# Plot no binding conditions
for i, sim in enumerate(no_binding_sims):
    label = 'No TIPOC Binding' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_no_binding, linestyle=':', linewidth=0.5)

# Plot non-local conditions
for i, sim in enumerate(non_local):
    label = 'No Local production' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_non_local, linestyle=':', linewidth=0.5)


plt.xlabel('Time (hours)')
plt.ylabel('Length (µm)')
plt.title('Total Hyphae Length Over Time')
plt.grid()
plt.legend()
plt.savefig('./outputdata/streptomyces_growthLog.png', dpi=300)
plt.yscale('log')
plt.show()
plt.clf()



# Plot Length vs Time ######### logarithmic scale with fit #########
color_normal = 'blue'
color_no_binding = 'red'
color_non_local = 'green'


plt.figure(figsize=(10,6))

# Plot normal conditions
for i, sim in enumerate(normal_sims):
    label = 'Normal Conditions' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_normal, linestyle=':', linewidth=0.5)

# Plot no binding conditions
for i, sim in enumerate(no_binding_sims):
    label = 'No TIPOC Binding' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_no_binding, linestyle=':', linewidth=0.5)

# Plot non-local conditions
for i, sim in enumerate(non_local):
    label = 'No Local production' if i == 0 else None
    plt.plot(sim[:,0], sim[:,1], label=label, color=color_non_local, linestyle=':', linewidth=0.5)

x_fit = np.linspace(initialTime, 8, 100)
# Normal Conditions Fit
y_fit_normal = np.exp(Mean_intercept_length_normal) * np.exp(Mean_slope_length_normal * x_fit)
plt.plot(x_fit, y_fit_normal, label='Normal Fit', color=color_normal, linestyle='--', linewidth=2)
# No Binding Fit
y_fit_no_binding = np.exp(Mean_intercept_length_no_binding) * np.exp(Mean_slope_length_no_binding * x_fit)
plt.plot(x_fit, y_fit_no_binding, label='No TIPOC Binding Fit', color=color_no_binding, linestyle='--', linewidth=2)
# No Local production Fit
y_fit_non_local = np.exp(Mean_intercept_length_non_local) * np.exp(Mean_slope_length_non_local * x_fit)
plt.plot(x_fit, y_fit_non_local, label='No Local production Fit', color=color_non_local, linestyle='--', linewidth=2)

plt.xlabel('Time (hours)')
plt.ylabel('Length (µm)')
plt.title('Total Hyphae Length Over Time')
plt.grid()
plt.legend()
plt.savefig('./outputdata/streptomyces_growthLog.png', dpi=300)
plt.yscale('log')
plt.show()
plt.clf()


# Plot Length vs Time ######### logarithmic scale only fit #########
color_normal = 'blue'
color_no_binding = 'red'
color_non_local = 'green'


plt.figure(figsize=(10,6))

x_fit = np.linspace(initialTime, 8, 100)
# Normal Conditions Fit
y_fit_normal = np.exp(Mean_intercept_length_normal) * np.exp(Mean_slope_length_normal * x_fit)
plt.plot(x_fit, y_fit_normal, label='Normal Fit', color=color_normal, linestyle='--', linewidth=2)
# No Binding Fit
y_fit_no_binding = np.exp(Mean_intercept_length_no_binding) * np.exp(Mean_slope_length_no_binding * x_fit)
plt.plot(x_fit, y_fit_no_binding, label='No TIPOC Binding Fit', color=color_no_binding, linestyle='--', linewidth=2)
# No Local production Fit
y_fit_non_local = np.exp(Mean_intercept_length_non_local) * np.exp(Mean_slope_length_non_local * x_fit)
plt.plot(x_fit, y_fit_non_local, label='No Local production Fit', color=color_non_local, linestyle='--', linewidth=2)


anova_test = stats.f_oneway(slopes_length_normal, slopes_length_no_binding, slopes_length_non_local)

plt.xlabel('Time (hours)')
plt.ylabel('Length (µm)')
plt.title('Total Hyphae Length Over Time')
plt.grid()
plt.legend()
plt.text(5, 40, f'ANOVA p-value: {anova_test.pvalue:.2}', fontsize=10)
plt.savefig('./outputdata/streptomyces_growthLog.png', dpi=300)
plt.yscale('log')
plt.show()
plt.clf()


# Plot Time vs Branches #########

plt.figure(figsize=(10,6))

# Plot normal conditions
for i, sim in enumerate(normal_sims):
    label = 'Normal Conditions' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_normal, linestyle=':', linewidth=0.5)

# Plot no binding conditions
for i, sim in enumerate(no_binding_sims):
    label = 'No TIPOC Binding' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_no_binding, linestyle=':', linewidth=0.5)

# Plot no local production conditions
for i, sim in enumerate(non_local):
    label = 'No Local production' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_non_local, linestyle=':', linewidth=0.5)


plt.xlabel('Time (hours)')
plt.ylabel('Branches')
plt.title('Number of Branches Over Time')
plt.grid()
plt.legend()
plt.savefig('./outputdata/streptomyces_branches.png', dpi=300)
plt.show()


# Plot Time vs Branches ######### logarithmic scale #########

plt.figure(figsize=(10,6))

# Plot normal conditions
for i, sim in enumerate(normal_sims):
    label = 'Normal Conditions' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_normal, linestyle=':', linewidth=0.5)

# Plot no binding conditions
for i, sim in enumerate(no_binding_sims):
    label = 'No TIPOC Binding' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_no_binding, linestyle=':', linewidth=0.5)

# Plot non-local conditions
for i, sim in enumerate(non_local):
    label = 'No Local production' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_non_local, linestyle=':', linewidth=0.5)


plt.xlabel('Time (hours)')
plt.ylabel('Branches')
plt.title('Number of Branches Over Time')
plt.grid()
plt.yscale('log')
plt.legend()
plt.savefig('./outputdata/streptomyces_branches.png', dpi=300)
plt.show()



# Plot Time vs Branches ######### logarithmic scale with fit #########

plt.figure(figsize=(10,6))

# Plot normal conditions
for i, sim in enumerate(normal_sims):
    label = 'Normal Conditions' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_normal, linestyle=':', linewidth=0.5)

# Plot no binding conditions
for i, sim in enumerate(no_binding_sims):
    label = 'No TIPOC Binding' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_no_binding, linestyle=':', linewidth=0.5)

# Plot non-local conditions
for i, sim in enumerate(non_local):
    label = 'No Local production' if i == 0 else None
    plt.plot(sim[:,0], sim[:,2], label=label, color=color_non_local, linestyle=':', linewidth=0.5)

x_fit = np.linspace(initialTime, 8, 100)
# Normal Conditions Fit
y_fit_normal = np.exp(Mean_intercept_branch_normal) * np.exp(Mean_slope_branch_normal * x_fit)
plt.plot(x_fit, y_fit_normal, label='Normal Fit', color=color_normal, linestyle='--', linewidth=2)
# No Binding Fit
y_fit_no_binding = np.exp(Mean_intercept_branch_no_binding) * np.exp(Mean_slope_branch_no_binding * x_fit)
plt.plot(x_fit, y_fit_no_binding, label='No Binding Fit', color=color_no_binding, linestyle='--', linewidth=2)
# Non-Local Fit
y_fit_non_local = np.exp(Mean_intercept_branch_non_local) * np.exp(Mean_slope_branch_non_local * x_fit)
plt.plot(x_fit, y_fit_non_local, label='Non-Local Fit', color=color_non_local, linestyle='--', linewidth=2)

plt.xlabel('Time (hours)')
plt.ylabel('Branches')
plt.title('Number of Branches Over Time')
plt.grid()
plt.yscale('log')
plt.legend()
plt.savefig('./outputdata/streptomyces_branches.png', dpi=300)
plt.show()


```

```python
## check normality of slopes_length_normal with qq plot and shapiro-wilk test
import scipy.stats as stats

stats.probplot(slopes_length_normal, dist="norm", plot=plt)
plt.title('QQ Plot of Length Growth Rates - Normal Conditions')
plt.savefig('./outputdata/qqplot_length_normal.png', dpi=300)
plt.show()
### Shapiro-Wilk test
shapiro_test = stats.shapiro(slopes_length_normal)
print(f'Shapiro-Wilk test for Length Growth Rates - Normal Conditions: W={shapiro_test.statistic}, p-value={shapiro_test.pvalue}')



stats.probplot(slopes_length_non_local, dist="norm", plot=plt)
plt.title('QQ Plot of Length Growth Rates - Non-local Conditions')
plt.savefig('./outputdata/qqplot_length_non_local.png', dpi=300)
plt.show()
### Shapiro-Wilk test
shapiro_test = stats.shapiro(slopes_length_non_local)
print(f'Shapiro-Wilk test for Length Growth Rates - Non-local Conditions: W={shapiro_test.statistic}, p-value={shapiro_test.pvalue}')  


stats.probplot(slopes_length_no_binding, dist="norm", plot=plt)
plt.title('QQ Plot of Length Growth Rates - No Binding Conditions')
plt.savefig('./outputdata/qqplot_length_no_binding.png', dpi=300)
plt.show()
### Shapiro-Wilk test
shapiro_test = stats.shapiro(slopes_length_no_binding)
print(f'Shapiro-Wilk test for Length Growth Rates - No Binding Conditions: W={shapiro_test.statistic}, p-value={shapiro_test.pvalue}')


## check equality of variances with Levene's test
levene_test = stats.levene(slopes_length_normal, slopes_length_no_binding, slopes_length_non_local)
print(f'Levene’s test for equality of variances: W={levene_test.statistic}, p-value={levene_test.pvalue}')
#if p-value > 0.05, we fail to reject the null hypothesis that the variances are equal. in
# other words, if p-value > 0.05, we can assume homogeneity of variances.


# ANOVA test to compare means of length growth rates across the three conditions
anova_test = stats.f_oneway(slopes_length_normal, slopes_length_no_binding, slopes_length_non_local)
print(f'ANOVA test for Length Growth Rates: F={anova_test.statistic}, p-value={anova_test.pvalue}')
# If p-value < 0.05, we reject the null hypothesis that all group means are equal.


```

```python

number_of_simulations = 13
initialTime=2
# Initialize lists to store the data
time_data = []
log_data = []
slopes = []
intercepts = []
# Loop through all the files

initial=int(initialTime*(60/1)*(60/1)*(1/5)) 

for i in range(0, number_of_simulations):
    # Generate the file name
    file_name = f'./inputdata/normalConditions/Simulation({i}).csv'
    
    # Load the data
    data = np.genfromtxt(file_name, delimiter=',', skip_header=0)
    
    # Transform the data
    data[:, 0] = data[:, 0] * (1/60) * (1/60)  # Convert seconds to hours
   

    # Fit a line to the data
    slope, intercept = np.polyfit(data[initial:, 0], np.log(data[initial:, 1]), 1)
    
    # Store the transformed data
    time_data.append(data[0:, 0])
    log_data.append(data[0:, 1])
    slopes.append(slope)
    intercepts.append(intercept)



# Calculate the average slope and intercept
average_slope = np.mean(slopes)
average_intercept = np.mean(intercepts)

# Plot the average line
t = np.linspace(initialTime, 8, 10000)  # 10 hours
u = np.e ** (average_slope*t + average_intercept)   

# Plot number of branches over time

# Plot each dataset
for i in range(0, number_of_simulations):
    plt.plot(time_data[i]+2, log_data[i], linestyle=':', color='gray', linewidth=1)

plt.plot(time_data[0]+2, log_data[0], linestyle=':', color='black', linewidth=0.7, label='Individual simulations')
plt.plot(t+2, u, color='black', linewidth=1.5, label=r'Growth rate (control)= {:.2f} $h^{{-1}}$'.format(average_slope))


plt.tick_params(axis='both', which='major', labelsize=12)
plt.legend(fontsize=12)
plt.xlabel('Time (hours)', fontsize=14)
plt.ylabel('Length [μm]', fontsize=14)
plt.yscale('log')
plt.xlim(5, 10)
plt.show()


########################### NO BINDING ##################################################################
number_of_simulations = 16
# Initialize lists to store the data
time_data = []
log_data = []
slopes = []
intercepts = []
# Loop through all the files

initial=int(initialTime*(60/1)*(60/1)*(1/5))

for i in range(0, number_of_simulations):
    # Generate the file name
    file_name = f'./inputdata/noBinding/Simulation({i}).csv'
    
    # Load the data
    data = np.genfromtxt(file_name, delimiter=',', skip_header=0)
    
    # Transform the data
    data[:, 0] = data[:, 0] * (1/60) * (1/60)  # Convert seconds to hours
   

    # Fit a line to the data
    slope, intercept = np.polyfit(data[initial:, 0], np.log(data[initial:, 1]), 1)
    
    # Store the transformed data
    time_data.append(data[0:, 0])
    log_data.append(data[0:, 1])
    slopes.append(slope)
    intercepts.append(intercept)



# Calculate the average slope and intercept
average_slope = np.mean(slopes)
average_intercept = np.mean(intercepts)

# Plot the average line
t = np.linspace(initialTime, 8, 10000)  # 10 hours
u = np.e ** (average_slope*t + average_intercept)   

# Plot number of branches over time

# Plot each dataset
for i in range(0, number_of_simulations):
    plt.plot(time_data[i]+2, log_data[i], linestyle=':', color='red', linewidth=0.5)

plt.plot(time_data[0]+2, log_data[0], linestyle=':', color='red', linewidth=0.7, label='Individual simulations')
plt.plot(t+2, u, color='red', linewidth=1.5, label=r'Growth rate (No Bind) = {:.2f} $h^{{-1}}$'.format(average_slope))


plt.tick_params(axis='both', which='major', labelsize=12)
plt.legend(fontsize=12)
plt.xlabel('Time (hours)', fontsize=14)
plt.ylabel('Length [μm]', fontsize=14)
plt.yscale('log')
plt.xlim(5, 10)


########################### NON local ##################################################################
number_of_simulations = 7
# Initialize lists to store the data
time_data = []
log_data = []
slopes = []
intercepts = []
# Loop through all the files

initial=int(initialTime*(60/1)*(60/1)*(1/5))

for i in range(0, number_of_simulations):
    # Generate the file name
    file_name = f'./inputdata/nonLocal/Simulation({i}).csv'
    
    # Load the data
    data = np.genfromtxt(file_name, delimiter=',', skip_header=0)
    
    # Transform the data
    data[:, 0] = data[:, 0] * (1/60) * (1/60)  # Convert seconds to hours
   

    # Fit a line to the data
    slope, intercept = np.polyfit(data[initial:, 0], np.log(data[initial:, 1]), 1)
    
    # Store the transformed data
    time_data.append(data[0:, 0])
    log_data.append(data[0:, 1])
    slopes.append(slope)
    intercepts.append(intercept)



# Calculate the average slope and intercept
average_slope = np.mean(slopes)
average_intercept = np.mean(intercepts)

# Plot the average line
t = np.linspace(initialTime, 8, 10000)  # 10 hours
u = np.e ** (average_slope*t + average_intercept)   

# Plot number of branches over time

# Plot each dataset
for i in range(0, number_of_simulations):
    plt.plot(time_data[i]+2, log_data[i], linestyle=':', color='green', linewidth=0.5)

plt.plot(time_data[0]+2, log_data[0], linestyle=':', color='green', linewidth=0.7, label='Individual simulations')
plt.plot(t+2, u, color='green', linewidth=1.5, label=r'Growth rate (Non-Local) = {:.2f} $h^{{-1}}$'.format(average_slope))


plt.tick_params(axis='both', which='major', labelsize=12)
plt.legend(fontsize=12)
plt.xlabel('Time (hours)', fontsize=14)
plt.ylabel('Length [μm]', fontsize=14)
plt.yscale('log')
plt.xlim(initialTime, 10)
plt.savefig('./outputdata/streptomyces_growth_comparison.png', dpi=300)
plt.show()
```
