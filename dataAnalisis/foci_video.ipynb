```python
import tifffile
import os

stack = tifffile.imread("./Spatiotemporal metabolism in Streptomyces/FOCI Dynamics - every 30 s/ROI_GFP.tif")  # shape: (41, 393, 468, 3)
os.makedirs("frames", exist_ok=True)
for i, frame in enumerate(stack):
    tifffile.imwrite(f"frames/frame_{i:03d}.tif", frame)
```

```python
ffmpeg -r 10 -i frames/frame_%03d.tif   -vcodec ffv1 -pix_fmt rgb24 GFP-8bit.mkv
```
